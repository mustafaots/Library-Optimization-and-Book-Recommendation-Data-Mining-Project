```python
import kagglehub        # for downloading datasets from Kaggle
import shutil           # for file operations like moving files
import os               # for operating system dependent functionality
```

```python
# this code snippet is from the kaggle webpage:

# Download the dataset (full dataset)
path = kagglehub.dataset_download("arashnic/book-recommendation-dataset")

print("Path to dataset files:", path)
```

```python
# source (downloaded folder)
source_folder = path

# destination (local project folder)
dest_folder = r"C:\Users\fafao\OneDrive\Desktop\Library-Optimization-and-Book-Recommendation-Data-Mining-Project\data\raw"

# verify destination exists
os.makedirs(dest_folder, exist_ok=True)

# copy each CSV file to the destination folder
for filename in ["Books.csv", "Users.csv", "Ratings.csv"]:
    shutil.copy(os.path.join(source_folder, filename),
                os.path.join(dest_folder, filename))

print("Files copied successfully!")
```
