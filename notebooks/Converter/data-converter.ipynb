```python
import pandas as pd

base_path = '../../data/raw/xlx/'
# Convert Excel files to CSV
excel_files = ['Borrowings.xlsx', 'library catalogue.xlsx', 'LIBRARY INVENTORY.xlsx']

excel_files = [base_path + file_name for file_name in excel_files]

for excel_file in excel_files:
    try:
        # Read Excel file
        df = pd.read_excel(excel_file)
        
        # Create CSV filename
        csv_file = excel_file.replace('.xlsx', '.csv')
        
        # Save as CSV
        df.to_csv(csv_file, index=False)
        
        print(f"Converted: {excel_file} -> {csv_file}")
        
    except Exception as e:
        print(f"Error converting {excel_file}: {e}")
```
