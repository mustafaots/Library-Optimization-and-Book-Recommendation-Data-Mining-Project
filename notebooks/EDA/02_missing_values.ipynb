```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
borrowings = pd.read_csv("../../data/processed/library_borrowings.csv")
catalogue = pd.read_csv("../../data/processed/library_catalogue.csv")
inventory = pd.read_csv("../../data/processed/library_inventory.csv")
```

```python
borrowings.isnull().sum().sort_values(ascending=False)
```

```python
catalogue.isnull().sum().sort_values(ascending=False)
```

```python
inventory.isnull().sum().sort_values(ascending=False)
```

```python
missing_catalogue = catalogue.isnull().mean() * 100
missing_catalogue.sort_values(ascending=False)
```

```python
missing_catalogue[missing_catalogue >= 0].plot(
    kind="bar",
    title="Missing Values Percentage (Catalogue)"
)
plt.ylabel("Percentage")
plt.show()
```
