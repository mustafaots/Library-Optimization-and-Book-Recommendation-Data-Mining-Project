```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
borrowings = pd.read_csv("../../data/processed/library_borrowings.csv")
catalogue = pd.read_csv("../../data/processed/library_catalogue.csv")
inventory = pd.read_csv("../../data/processed/library_inventory.csv")
```

```python
borrowings['borrowing_duration'].hist(bins=30)
plt.title("Distribution of Borrowing Duration")
plt.xlabel("Days")
plt.ylabel("Frequency")
plt.show()
```

```python
borrowings['Title'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Most Borrowed Books")
plt.ylabel("Borrow Count")
plt.show()
```

```python
borrowings['category'].value_counts().plot(kind='bar')
plt.title("Borrowings by Category")
plt.ylabel("Number of Borrowings")
plt.show()
```

```python
catalogue['Copy Count'].hist(bins=20)
plt.title("Distribution of Copies per Book in Catalogue")
plt.xlabel("Number of Copies")
plt.ylabel("Number of Books")
plt.show()
```

```python
inventory['Status'].value_counts().plot(kind='bar')
plt.title("Inventory Status Distribution")
plt.ylabel("Number of Books")
plt.show()
```

```python
inventory['Acquisition Mode'].value_counts().plot(kind='bar')
plt.title("Inventory Acquisition Modes")
plt.ylabel("Number of Books")
plt.show()
```
